```python
import json
```

* `json` 패키지: 직렬화(serialization), 역직렬화(de-serialization) 기능을 가지고 있는 패키지.
* 직렬화: 객체(예: 딕셔너리) --> JSON 형식의 문자열
* 역직렬화: JSON 형식 문자열 --> 객체(예: 딕셔너리)
* JSON(JavaScript Object Notation)
  * 자바스크립트 언어의 객체 표현 방식.
  * 속성(property)과 속성의 값으로 객체들을 표현.
  * 자바스크립트의 객체 표현법은 파이썬의 딕셔너리 표현법(key: value)과 매우 유사.
* `json.loads(str)`: JSON 형식으로 작성된 문자열을 파이썬 dict로 변환해서 리턴. 역직렬화 함수.
* `json.dumps(object)`: 객체(딕셔너리)를 JSON 형식의 문자열로 변환해서 리턴. 직렬화 함수.

```python
# Python dict 객체
student  = {
    'id': 1,
    'name': '오쌤',
    'age': 16
}
```

```python
student
```

```python
# dict --> str
student_json = json.dumps(student)
```

```python
student_json
```

```python
# str --> dict
result = json.loads(student_json)
```

```python
result
```

```python
arguments = "{\"timezone\":\"Asia/Seoul\"}"
arguments
```

```python
args = json.loads(arguments)  # de-serialization(역직렬화): 문자열 -> 딕셔너리
args
```
