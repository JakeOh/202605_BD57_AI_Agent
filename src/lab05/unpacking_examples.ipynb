# Unpacking Exmaples

## 1. Unpacking 기본

unpacking: list, tuple, dict와 같은 iterable의 묶여 있는 값들을 분해해서 변수에 할당하는 것.

```python
name, age = ('오쌤', 16)
```

```python
name
```

```python
age
```

## 2. dict의 unpacking

```python
person = {'name': '오쌤', 'age': 16}
```

```python
x, y = person
```

```python
x
```

```python
y
```

변수 x, y에는 dict의 key가 할당됨.

```python
x, y = person.values()
```

```python
x
```

```python
y
```

```python
x, y = person.items()
```

```python
x  #> (key1, value1) 튜플
```

```python
y  #> (key2, value2) 튜플
```

## 3. 중첩 구조 unpacking

```python
name, (year, month, day) = ['오쌤', (2026, 10, 2)]
```

```python
name
```

```python
year
```

```python
month
```

```python
day
```

```python
(key1, value1), (key2, value2) = person.items()
```

```python
print(key1)
print(value1)
print(key2)
print(value2)
```

```python
x
```
